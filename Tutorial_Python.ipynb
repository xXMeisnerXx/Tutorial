## 1. Variables y tipos de datos

Una **variable** es un nombre que guarda un valor: `nombre = valor`.

Reglas: solo letras, números y `_`; no puede empezar con un número; `edad` y `Edad` son variables distintas; no se pueden usar palabras reservadas como `if` o `for`. Usa nombres que digan qué contienen.

### `int` y `float`: números
`int` guarda enteros y `float` guarda decimales. El `_` dentro de un número solo sirve para leerlo mejor.

```python
edad = 25                 # int
salario = 2_500_000.0     # float
```

### `print()`: muestra valores
En Jupyter, la última línea de una celda se muestra sola. Para ver varios valores, usa `print()` en cada uno.

```python
print(edad)
print(salario)
```

### f-strings: texto con variables
Pon una `f` antes de las comillas y escribe las variables entre `{}`.

```python
print(f"Tengo {edad} años y gano {salario} pesos")
```

### `str`: texto
Se escribe entre comillas simples o dobles; las dos funcionan igual.

```python
ciudad = "Medellín"
pais = 'Colombia'
print(ciudad, pais)
```

### `bool`: verdadero o falso
Solo tiene dos valores: `True` o `False`.

```python
es_estudiante = True
tiene_deudas = False
print(es_estudiante, tiene_deudas)
```

### `type()`: consulta el tipo de un dato

```python
print(type(edad))
print(type(salario))
print(type(ciudad))
print(type(es_estudiante))
```

### `del`: borra una variable
Después de borrarla, usarla produce un error porque ya no existe.

```python
temporal = 100
del temporal
# print(temporal)   # Quita el # para ver el error: NameError
```

## 2. Operaciones

### Operaciones aritméticas
La división `/` siempre devuelve un `float`.

```python
precio = 12_000
cantidad = 3

print(precio + 500)        # suma
print(precio - 2_000)      # resta
print(precio * cantidad)   # multiplicación
print(precio / cantidad)   # división
print(2 ** 3)              # potencia
print(17 // 5)             # división entera
print(17 % 5)              # residuo
```

### Comparaciones: `>`, `>=`, `<`, `<=`, `==`, `!=`
Devuelven `True` o `False`. Ojo: `=` asigna un valor y `==` compara.

```python
print(edad >= 18)               # mayor o igual
print(salario < 1_000_000)      # menor que
print(ciudad == "Medellín")     # igual a
print(ciudad != "Bogotá")       # diferente de
```

### Operadores lógicos: `and`, `or`, `not`
- `and`: las dos condiciones deben ser verdaderas.
- `or`: basta con que una lo sea.
- `not`: invierte el valor.

**Ojo:** los símbolos `&`, `|` y `~` no son lo mismo. Son operadores de bits y se usan más adelante en pandas. En Python normal, `~True` da `-2`, no `False`.

```python
print((edad < 30) and es_estudiante)   # True: se cumplen las dos
print(tiene_deudas or es_estudiante)   # True: se cumple una
print(not es_estudiante)               # False
print(~True)                           # -2: por eso no se usa ~ en lugar de not
```

### `isinstance()`: ¿el dato es de este tipo?

```python
print(isinstance(edad, int))       # True
print(isinstance(ciudad, float))   # False
```

### Concatenar textos con `+`

```python
nombre_completo = "Juan" + "_" + "Pérez"
nombre_completo
```

## 3. Conversión de tipos

### `str()`: convierte a texto

```python
celular = 3001234567
celular_texto = str(celular)
print(celular_texto, type(celular_texto))
```

### `int()`: convierte a entero
Corta los decimales, **no redondea**. También convierte texto, si el texto es un número.

```python
print(int(4.9))      # 4, no 5
print(int("300"))    # texto numérico a entero
print(int(True))     # True vale 1
# int("hola")        # Quita el # para ver el error: ValueError
```

### `float()`: convierte a decimal

```python
precio_texto = "19.99"
float(precio_texto)
```

## 4. Métodos de texto

Una **clase** es una plantilla: por ejemplo, `str`. Un **objeto** es un valor creado con esa plantilla, como `"hola"`.

Los **métodos** son acciones del objeto (verbos) y se escriben `objeto.metodo()`. Los **atributos** son cualidades y van sin paréntesis: `objeto.atributo`.

### `.upper()`: todo en mayúsculas

```python
"medellín".upper()
```

### `.lower()`: todo en minúsculas

```python
"ANTIOQUIA".lower()
```

### `.title()`: primera letra de cada palabra en mayúscula

```python
"universidad autónoma latinoamericana".title()
```

### `.capitalize()`: solo la primera letra en mayúscula

```python
"bELLO es un municipio".capitalize()
```

### `.startswith()`: ¿empieza con...?

```python
"COD-1234".startswith("COD")
```

### `.endswith()`: ¿termina con...?

```python
"informe_ventas.xlsx".endswith(".xlsx")
```

### `.isalpha()`: ¿solo tiene letras? Espacios y números hacen que sea `False`.

```python
print("Envigado".isalpha())        # True
print("Envigado 2026".isalpha())   # False
```

### `.isnumeric()`: ¿solo tiene números?

```python
"2026".isnumeric()
```

### `.replace()`: reemplaza una parte del texto

```python
"1.250.000".replace(".", "")   # quita los puntos de miles
```

### `.split()`: parte el texto en una lista

```python
"Medellín,Bello,Itagüí".split(",")
```

### `.strip()`: quita espacios al inicio y al final

```python
"   hola   ".strip()
```

## 5. Tuplas

Colección **ordenada e inmutable**: una vez creada, no se puede modificar. Se escribe con `()`.

### Crear una tupla

```python
municipios = ("Medellín", "Bello", "Itagüí", "Envigado", "Sabaneta")
vacia = ()   # tupla vacía; también sirve tuple()
municipios
```

### `.count()`: cuenta cuántas veces aparece un valor

```python
atendidos = ("Juan", "María", "Carlos", "Juan", "Luis")
atendidos.count("Juan")
```

### Desempaquetar: asigna cada elemento a una variable

```python
datos_cliente = ("Juan", "Ramírez", 3001234567)
nombre, apellido, telefono = datos_cliente
print(nombre)
print(apellido)
print(telefono)
```

### `len()`: cuenta los elementos (sirve para cualquier colección)

```python
len(datos_cliente)
```

## 6. Listas

Colección **ordenada y mutable**: se puede modificar después de crearla. Se escribe con `[]`.

Varios métodos modifican la lista **en el lugar**: cambian la original y no crean una nueva.

### Crear una lista

```python
clientes = ["Ana", "Luis", "María", "Carlos", "Luis"]
vacia = []   # lista vacía; también sirve list()
clientes
```

### `.append()`: agrega un elemento al final
Si le pasas una lista, la agrega como **un solo** elemento.

```python
clientes.append("Andrés")
clientes
```

### `.remove()`: borra la primera coincidencia del valor

```python
clientes.remove("Luis")
clientes
```

### `+`: une dos listas en una **nueva**

```python
fin_de_semana = ["Pedro", "Lucía"]
todos = clientes + fin_de_semana
todos
```

### `.extend()`: agrega los elementos de otra lista a la original

```python
clientes.extend(["Sofía", "Diego"])
clientes
```

### `.insert()`: agrega un elemento en una posición (se cuenta desde 0)

```python
clientes.insert(1, "Camila")   # posición 1 = segundo lugar
clientes
```

### `.pop()`: quita el elemento de una posición y lo devuelve

```python
quitado = clientes.pop(1)
print(quitado)
print(clientes)
```

### `.sort()`: ordena de menor a mayor (o alfabéticamente)

```python
precios = [5_000, 1_200, 8_900, 3_400]
precios.sort()
precios
```

### `.sort(reverse=True)`: ordena de mayor a menor

```python
precios.sort(reverse=True)
precios
```

### Cuidado: `=` no copia una lista
Con `=`, los dos nombres apuntan a **la misma** lista. Si cambias una, cambia la otra.

```python
dias = ["lunes", "martes"]
dias_2 = dias              # no es una copia
dias_2.append("miércoles")
print(dias)                # la original también cambió
```

### `.copy()`: crea una copia independiente

```python
dias_3 = dias.copy()
dias_3.append("jueves")
print(dias)     # no cambia
print(dias_3)
```

## 7. Conjuntos (`set`)

Colección **mutable, sin orden y sin repetidos**. Se escribe con `{}`. Para crear uno vacío usa `set()`, porque `{}` crea un diccionario.

### Crear un conjunto: los repetidos se eliminan solos

```python
marcas = {"Marca_A", "Marca_C", "Marca_B", "Marca_A"}
marcas
```

### `set()`: quitar duplicados de una lista

```python
vendidos = ["Pan", "Leche", "Pan", "Huevos", "Leche"]
list(set(vendidos))   # el orden puede variar: los conjuntos no tienen orden
```

### `.add()`: agrega un elemento

```python
marcas.add("Marca_D")
marcas
```

### `.intersection()`: elementos que están en los dos

```python
inventario_ayer = {"ID1234", "ID5678", "ID9012"}
inventario_hoy = {"ID5678", "ID9012", "ID3456"}
inventario_ayer.intersection(inventario_hoy)   # atajo: inventario_ayer & inventario_hoy
```

### `.union()`: todos los elementos, sin repetir

```python
inventario_ayer.union(inventario_hoy)   # atajo: inventario_ayer | inventario_hoy
```

### `.difference()`: los que están en el primero pero no en el segundo

```python
inventario_ayer.difference(inventario_hoy)   # atajo: inventario_ayer - inventario_hoy
```

## 8. Diccionarios

Guardan pares **clave: valor**. Las claves no se repiten y los valores pueden ser cualquier cosa.

Desde Python 3.7 conservan el orden en que se agregaron las claves, pero se consultan por clave, no por posición.

### Crear con `{}`

```python
cliente = {"nombre": "María", "edad": 30, "ciudad": "Bello"}
cliente
```

### Crear con `dict()`: las claves se escriben sin comillas

```python
producto = dict(codigo="P001", nombre="Café", precio=15_000)
producto
```

### Valores que son listas u otros diccionarios (anidados)

```python
personas = {
    "ID1": {"nombre": "Juan", "ciudad": "Medellín"},
    "ID2": {"nombre": "María", "ciudad": "Bello", "habilidades": ["python", "sql"]},
}
personas
```

### `.keys()`: todas las claves

```python
empleado = {"nombre": "Valeria", "apellido": "Rojas", "habilidades": ["python", "sql", "excel"]}
empleado.keys()
```

### `.values()`: todos los valores

```python
empleado.values()
```

### `.items()`: los pares (clave, valor)

```python
empleado.items()
```

### `.get()`: busca una clave; si no existe, devuelve un valor por defecto en vez de un error

```python
print(empleado.get("nombre", "sin dato"))
print(empleado.get("area", "sin dato"))   # la clave no existe
```

### `.update()`: agrega o cambia varios pares a la vez

```python
empleado.update({"area": "analítica", "turno": "diurno"})
empleado
```

### `.pop()`: quita una clave y devuelve su valor

```python
empleado.pop("turno")
empleado
```

## 9. Indexación y slicing

- Las posiciones empiezan en **0**.
- Los índices negativos cuentan desde el final.
- Los conjuntos no se pueden indexar, porque no tienen orden.
- En los diccionarios se usa la clave, no la posición.

```python
letras = ["a", "b", "c", "d", "e", "f"]
vocales = ("a", "e", "i", "o", "u")
```

### `[i]`: una posición

```python
print(letras[0])    # primera
print(vocales[1])   # segunda
# letras[10]        # Quita el # para ver el error: IndexError, la posición no existe
```

### `[-i]`: desde el final

```python
print(letras[-1])   # última
print(letras[-2])   # penúltima
```

### `[clave]`: en diccionarios (y diccionarios anidados)

```python
print(cliente["ciudad"])
print(personas["ID2"]["ciudad"])   # primero el ID y luego el dato
```

### `[inicio:fin]`: un rango
La posición `fin` **no** se incluye.

```python
letras[0:3]   # posiciones 0, 1 y 2
```

### `[:fin]` y `[inicio:]`: omitir el inicio o el final

```python
print(letras[:2])   # desde el inicio hasta la posición 1
print(letras[2:])   # desde la posición 2 hasta el final
```

### `[::paso]`: saltar de a varios

```python
letras[::2]   # de dos en dos
```

### `[::-1]`: invertir el orden

```python
letras[::-1]
```

### Modificar una posición: `estructura[posicion] = nuevo_valor`

```python
letras[-1] = "z"
letras
```

### Modificar un diccionario por clave

```python
cliente["edad"] = 31
cliente
```

### Las tuplas no se pueden modificar

```python
# vocales[0] = "A"   # Quita el # para ver el error: TypeError
```

### Modificar un rango
El lado derecho debe ser una colección, como una lista. Si es un número solo, da `TypeError`. La lista se ajusta si el tamaño es distinto.

```python
numeros = [10, 20, 30, 40, 50]
numeros[0:3] = [1, 2, 3]
numeros
```
